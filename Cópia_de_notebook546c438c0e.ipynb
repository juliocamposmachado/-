```python
#!/usr/bin/env python3
"""
Setup script for Red-Teaming Challenge - OpenAI GPT-OSS-20b
Author: Competition Participant
License: CC0 (as required by competition rules)
"""

import subprocess
import sys
import os

def install_dependencies():
    """Install required dependencies for red-teaming gpt-oss-20b"""

    print("🔧 Installing PyTorch and CUDA support...")
    subprocess.run([
        sys.executable, "-m", "pip", "install", "-q", "--upgrade", "torch",
        "--index-url", "https://download.pytorch.org/whl/cu128"
    ])

    print("🔧 Installing Transformers and Triton kernels...")
    subprocess.run([
        sys.executable, "-m", "pip", "install", "-q",
        "git+https://github.com/huggingface/transformers",
        "triton==3.4",
        "git+https://github.com/triton-lang/triton.git@main#subdirectory=python/triton_kernels"
    ])

    print("🔧 Removing conflicting dependencies...")
    subprocess.run([
        sys.executable, "-m", "pip", "uninstall", "-q", "torchvision", "torchaudio", "-y"
    ])

    print("🔧 Installing additional tools for red-teaming...")
    subprocess.run([
        sys.executable, "-m", "pip", "install", "-q",
        "datasets",
        "evaluate",
        "wandb",
        "tensorboard"
    ])

def setup_model():
    """Load and setup gpt-oss-20b model for testing"""
    from transformers import AutoModelForCausalLM, AutoTokenizer

    print("🤖 Loading gpt-oss-20b model...")
    model_id = "openai/gpt-oss-20b"

    tokenizer = AutoTokenizer.from_pretrained(model_id)
    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        torch_dtype="auto",
        device_map="auto",  # Use "cuda" if you have specific GPU preferences
        trust_remote_code=True
    )

    return model, tokenizer

def test_basic_functionality(model, tokenizer):
    """Test basic model functionality before red-teaming"""

    print("🧪 Testing basic model functionality...")

    messages = [
        {"role": "system", "content": "You are a helpful AI assistant."},
        {"role": "user", "content": "Hello, can you help me test this system?"},
    ]

    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        return_tensors="pt",
        return_dict=True,
    ).to(model.device)

    generated = model.generate(**inputs, max_new_tokens=100, do_sample=True, temperature=0.7)
    response = tokenizer.decode(generated[0][inputs["input_ids"].shape[-1]:])

    print("✅ Model response:", response)
    return True

def main():
    """Main setup function"""
    print("🚨 RED-TEAMING CHALLENGE - OpenAI GPT-OSS-20b Setup")
    print("⚖️  Licensed under CC0 as required by competition rules")
    print("🎯 Competition: https://www.kaggle.com/competitions/openai-gpt-oss-20b-red-teaming")
    print("-" * 80)

    try:
        install_dependencies()
        model, tokenizer = setup_model()
        test_basic_functionality(model, tokenizer)

        print("\n✅ Setup completed successfully!")
        print("🔴 Ready to begin red-teaming activities")
        print("\n⚠️  IMPORTANT REMINDERS:")
        print("   - This is for research/safety purposes only")
        print("   - Follow OpenAI's usage policies")
        print("   - Document all findings responsibly")
        print("   - Submit results via Kaggle platform")

    except Exception as e:
        print(f"❌ Setup failed: {e}")
        sys.exit(1)

if __name__ == "__main__":
    main()
```
